```python
import re
import string
```

Como import re podemos limpiar nuestro texto de caracteres especiales y puntuación a traves de un patrón que encapsule todos los caracteres que no queremos o que en otro caso queremos encontrar.

```python
print(string.punctuation)  
```

```python
re_punc = re.compile('[%s]' % re.escape(string.punctuation))
```

```python
stripped = [re_punc.sub('', w) for w in words]
```

```python
print(stripped[:100])
```

Convierte a minúsculas el texto, utiliza lower(), lee cómo funciona este método y

Ahora, todo el texto esta en minusculas, no cambio nada mas.

```python
words_lower = [w.lower() for w in stripped]
```

w.lower converte todas las palabras de nuestro corpues a minúsculas

```python
print(words_lower[:100])
```

-------------------------------------------------------------------------------------------------------------

```python
from nltk.corpus import stopwords 
```

```python
stop_words = stopwords.words('english')
```

```python
primeras_10 = stop_words[:10]
```

```python
print(primeras_10)
```

```python
filtrar_stopwords = [w for w in words_lower if w not in stop_words and w != '']
```

```python
print(filtrar_stopwords[:100])
```

-------------------------------------------------------------------------------------------------------------

```python
from nltk.stem import SnowballStemmer
```

```python
truncar = SnowballStemmer('english')
```

```python
truncar_words = [truncar.stem(w) for w in filtrar_stopwords]
```

```python
print(truncar_words[:100])
```

-------------------------------------------------------------------------------------------------------------

-------------------------------------------------------------------------------------------------------------

```python
from nltk.stem import WordNetLemmatizer
```

```python
wnl = WordNetLemmatizer()
```

```python
lematizar = [wnl.lemmatize(w) for w in filtrar_stopwords]
```

```python
print(lematizar[:100])
```

Con snowball solo truncamos las palabras a su raíz, por ejemplo, la palabra "running" se truncará a "run". Esto nos ayuda a reducir la variabilidad de las palabras y a enfocarnos en su significado básico.

Con lematización hacemos un analisis morfológico de las palabras para reducirlas a su forma base o lema. Por ejemplo, la palabra "better" se lematizará a "good". Esto nos permite agrupar palabras que tienen el mismo significado pero diferentes formas gramaticales.
