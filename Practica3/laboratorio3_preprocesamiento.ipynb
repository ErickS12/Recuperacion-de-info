LABORATORIO 3 - Preprocesamiento

Toma consultas_sin_preprocesar.txt (formato: Numero | titulo | texto) y limpia
unicamente el titulo y el texto, conservando intacto el numero de indice.

Pasos: separar palabras compuestas (- / _), quitar puntuacion, minusculas,
quitar numeros, palabras con numeros y letras sueltas, quitar stopwords
y truncar con Snowball.

```python
import os
```

```python
import re
```

```python
import string
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.stem import SnowballStemmer
```

```python
corpus_root = r"C:\Users\erick\OneDrive\Documentos\documentos UNI\Semestre 8\Recuperacion de info\Practicas en clase\Practica3"
```

```python
archivo_entrada = os.path.join(corpus_root, "consultas_sin_preprocesar.txt")
```

```python
with open(archivo_entrada, encoding="utf-8") as f:
    lineas = f.read().splitlines()
```

```python
print(lineas[:50])
```

```python
re_punc = re.compile('[%s]' % re.escape(string.punctuation))
```

```python
stop_words = stopwords.words('english')
```

```python
truncar = SnowballStemmer('english')
```

```python
#este for recorre cada linea del archivo de entrada, separa el numero de la consulta, el titulo y el texto, luego limpia el titulo y el texto de caracteres especiales, separa las palabras, elimina palabras cortas, numeros y stopwords, y finalmente aplica stemming a las palabras restantes. El resultado se guarda en una lista de lineas de salida.

lineas_salida = []
for linea in lineas:
    num, titulo, texto = linea.split(" | ")
    limpio = re.sub(r"[-/_]", " ", titulo + " " + texto)     # separa palabras compuestas
    palabras = limpio.split() # separa las palabras
    palabras = [re_punc.sub('', w).lower() for w in palabras] # quita la puntuacion y pasa a minusculas
    palabras = [w for w in palabras if w.isalpha() and len(w) > 1]   # fuera numeros, palabras con numeros y letras sueltas
    palabras = [w for w in palabras if w not in stop_words]
    palabras = [truncar.stem(w) for w in palabras]
    lineas_salida.append(f"{num} | {' '.join(palabras)}")
```

```python
print(lineas_salida[:50])
```

```python
archivo_salida = os.path.join(corpus_root, "consultas_preprocesadas.txt")
```

```python
with open(archivo_salida, "w", encoding="utf-8") as f:
    f.write("\n".join(lineas_salida))
```
