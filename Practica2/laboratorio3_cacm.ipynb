Generar los archivos de documentos y de consultas a partir de cacm.all y
query.text con el formato:

Numero | titulo | texto

```python
import os
```

```python
import re
```

```python
corpus_root = r"C:\Users\erick\OneDrive\Documentos\documentos UNI\Semestre 8\Recuperacion de info\Practicas en clase\Practica3"
```

DOCUMENTOS (cacm.all)

```python
archivo = os.path.join(corpus_root, "cacm.all")
```

```python
with open(archivo, encoding="utf-8", errors="ignore") as f:
    lineas = f.read().splitlines()
```

```python
print(lineas[:10])
```

```python
registros = []
```

```python
actual = None
```

```python
campo = None
```

```python
for linea in lineas:
    if re.match(r"^\.[A-Z]", linea):
        campo = linea[1]
        if campo == "I":
            actual = {"id": linea[2:].strip(), "T": [], "W": []}
            registros.append(actual)
    elif actual is not None and campo in ("T", "W"):
        actual[campo].append(linea.strip())
```

```python
print(len(registros), "registros encontrados")
```

```python
print(registros[19])
```

```python
lineas_salida = []
for r in registros:
    titulo = re.sub(r"\s+", " ", " ".join(r["T"]).replace("|", " ")).strip()
    texto = re.sub(r"\s+", " ", " ".join(r["W"]).replace("|", " ")).strip()
    lineas_salida.append(f"{r['id']} | {titulo} | {texto}")
```

```python
print(lineas_salida[19])
```

```python
archivo_salida = os.path.join(corpus_root, "documentos_sin_preprocesar.txt")
```

```python
with open(archivo_salida, "w", encoding="utf-8") as f:
    f.write("\n".join(lineas_salida))
```

```python
print(len(lineas_salida), "lineas ->", archivo_salida)
```

CONSULTAS (query.text)

El mismo codigo, cambiando el archivo de entrada y el de salida. query.text no

```python
archivo = os.path.join(corpus_root, "query.text")
```

```python
print(lineas[:10])
```

```python
registros = []
actual = None
campo = None
```

```python
for linea in lineas:
    if re.match(r"^\.[A-Z]", linea):
        campo = linea[1]
        if campo == "I":
            actual = {"id": linea[2:].strip(), "T": [], "W": []}
            registros.append(actual)
    elif actual is not None and campo in ("T", "W"):
        actual[campo].append(linea.strip())

registros = [r for r in registros if r["T"] or r["W"]]   # CORREGIDO: query.text termina con un ".I 0" vacio
```
