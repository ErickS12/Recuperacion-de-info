LABORATORIO 3 - Preprocesamiento

Toma documentos_sin_preprocesar.txt y consultas_sin_preprocesar.txt
(formato: Numero | titulo | texto) y limpia unicamente el titulo y el texto,
conservando intacto el numero de indice y los dos separadores.

Pasos: separar palabras compuestas (- / _), quitar puntuacion, minusculas,
quitar numeros, palabras con numeros y letras sueltas, quitar stopwords
y truncar con Snowball.

```python
import os
```

```python
import re
```

```python
import string
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.stem import SnowballStemmer
```

```python
corpus_root = r"C:\Users\erick\OneDrive\Documentos\documentos UNI\Semestre 8\Recuperacion de info\Practicas en clase\Practica2"
```

```python
re_punc = re.compile('[%s]' % re.escape(string.punctuation))
```

```python
stop_words = stopwords.words('english')
```

```python
truncar = SnowballStemmer('english')
```

```python
#este for recorre cada linea del archivo de entrada, separa el numero de la consulta, el titulo y el texto, luego limpia el titulo y el texto de caracteres especiales, separa las palabras, elimina palabras cortas, numeros y stopwords, y finalmente aplica stemming a las palabras restantes. El resultado se guarda en una lista de lineas de salida.

def preprocesar_archivo(nombre_entrada, nombre_salida):
    with open(os.path.join(corpus_root, nombre_entrada), encoding="utf-8") as f:
        lineas = f.read().splitlines()

    lineas_salida = []
    for linea in lineas:
        num, titulo, texto = linea.split(" | ")
        campos = []
        for campo in (titulo, texto):                                      # CORREGIDO: titulo y texto por separado
            limpio = re.sub(r"[-/_]", " ", campo)                          # separa palabras compuestas
            palabras = limpio.split()                                      # separa las palabras
            palabras = [re_punc.sub('', w).lower() for w in palabras]      # quita la puntuacion y pasa a minusculas
            palabras = [w for w in palabras if w.isalpha() and len(w) > 1] # fuera numeros, palabras con numeros y letras sueltas
            palabras = [w for w in palabras if w not in stop_words]
            palabras = [truncar.stem(w) for w in palabras]
            campos.append(' '.join(palabras))
        lineas_salida.append(f"{num} | {campos[0]} | {campos[1]}")         # CORREGIDO: se conserva el segundo "|"

    with open(os.path.join(corpus_root, nombre_salida), "w", encoding="utf-8") as f:
        f.write("\n".join(lineas_salida))

    print(len(lineas_salida), "lineas ->", nombre_salida)
    return lineas_salida
```

DOCUMENTOS

```python
docs_preprocesados = preprocesar_archivo("documentos_sin_preprocesar.txt", "documentos_preprocesados.txt")
```
